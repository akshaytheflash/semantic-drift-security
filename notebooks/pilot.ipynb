# Pilot

Reproduces the pilot report from the recorded JSONL. No API calls.

## What the pilot established (see `docs/pilot_report.md`)

- With 9 probes at 0-100 interactions (<=~11k tokens) no arm showed a rising violation rate on `glm-5.3-flash`.
- Six of nine probes never fired; P11 was mis-specified and moved to strict mode.
- The horizon was short in tokens, so a large-volume arm set (`S~*`, ~30k tokens) was added before the main run.

import sys, json
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()))
RAW = sorted(Path('experiments/raw').glob('*pilot_v3*.jsonl'))[0]
print('raw file:', RAW)

recs = [json.loads(l) for l in open(RAW)]
print(len(recs), 'checkpoint records')
for r in recs:
    v = sum(o['violated'] for o in r['probes'].values() if o['kind']=='adversarial')
    print(f"{r['group']:2} cp={r['checkpoint']:>3} items={r['n_context_items']:>3} tok={r['history_tokens']:>6} viol={v}/9")
